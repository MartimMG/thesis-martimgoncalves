```python
import pandas as pd
```

```python
dataset = pd.read_csv('../../../Breast Cancer METABRIC.csv')
```

```python
dataset.columns
```

```python
dataset = dataset[dataset['Cancer Type'] == 'Breast Cancer']
```

```python
# compare 'ER status measured by IHC' and 'ER Status' for differences
col1 = 'ER status measured by IHC'
col2 = 'ER Status'

s1 = dataset[col1].astype(str).str.strip().str.lower()
s2 = dataset[col2].astype(str).str.strip().str.lower()

isna1 = dataset[col1].isna()
isna2 = dataset[col2].isna()

both_nonmissing_diff = (~isna1 & ~isna2) & (s1 != s2)
one_missing = isna1 ^ isna2

print("Total rows:", len(dataset))
print("Both non-missing & different:", both_nonmissing_diff.sum())
print("One missing & other present:", one_missing.sum())

# DataFrame of mismatches (including cases where one is missing)
mismatch_df = dataset[both_nonmissing_diff | one_missing].copy()
mismatch_df['_ER_IHC_norm'] = s1
mismatch_df['_ER_Status_norm'] = s2

# show first mismatches
mismatch_df[['Patient ID', col1, col2, '_ER_IHC_norm', '_ER_Status_norm']].head(50)
```
